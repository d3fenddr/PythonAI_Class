# Lesson 03 ? Creative Text Generation with Markov Chains

This lesson demystifies how machines improvise text. We'll build a character-level Markov generator that morphs source material into new, quirky sentences.

## Learning Goals
- Translate text into Markov state transitions
- Sample characters probabilistically to create fresh sequences
- Evaluate creativity by comparing multiple generated samples

```python
import random
from collections import defaultdict, Counter

random.seed(1337)
```

## Step 1 ? Choose a Voice
We'll use a mash-up of sci-fi and philosophical quotes as the seed corpus. Feel free to replace the text with your own favorite material.

```python
corpus = """
The future is already here ? it's just not evenly distributed.
We are all born ignorant, but one must work hard to remain stupid.
Reality is that which, when you stop believing in it, doesn't go away.
The only way to discover the limits of the possible is to go beyond them into the impossible.
Consciousness is only possible through change; change is only possible through movement.
""".strip()

print(corpus)
```

## Step 2 ? Build the Markov Model
A Markov model learns the probability distribution of the next character given the current one (or many). We'll use 3-character states to capture short phrases.

```python
def build_markov_chain(text: str, order: int = 3):
    chain = defaultdict(Counter)
    padded = "~" * order + text

    for idx in range(len(text)):
        state = padded[idx:idx + order]
        next_char = text[idx]
        chain[state][next_char] += 1

    return chain


chain = build_markov_chain(corpus, order=3)
len(chain)
```

### Sampling Strategy
We'll repeatedly sample the next character using the transition probabilities until we hit a stopping condition.

```python
def sample_next(counter: Counter):
    total = sum(counter.values())
    threshold = random.uniform(0, total)
    cumulative = 0
    for char, count in counter.items():
        cumulative += count
        if cumulative >= threshold:
            return char
    return random.choice(list(counter.keys()))


def generate_text(chain, length=240, order=3):
    state = "~" * order
    generated = []

    for _ in range(length):
        counter = chain.get(state)
        if not counter:
            break
        next_char = sample_next(counter)
        generated.append(next_char)
        state = state[1:] + next_char

    return "".join(generated)


generate_text(chain)
```

## Step 3 ? Compare Outputs
Let's generate a few samples with different random seeds to see the spectrum of creativity.

```python
for seed in range(3):
    random.seed(seed)
    print(f"--- Sample {seed + 1} ---")
    print(generate_text(chain, length=220))
    print()
```

### Analyze the Flavor
Some outputs will sound coherent, others will glitch. That's a Markov chain letting local probabilities run wild. Increase the order to tighten grammar, or feed in new corpora to change the vibe.

## Wrap-Up
Generative AI often reduces to clever probabilistic sampling. By building it yourself you gain intuition for temperature, context windows, and why large language models feel so fluid. Next, try word-level transitions or beam-search sampling to steer the narrative.
